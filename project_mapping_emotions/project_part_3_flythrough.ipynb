## How the Flythrough Works

`flythrough_config.js` defines an array of **chapters**. Each chapter is one "stop" on the map tour:

```json
{
    "id": "stop-1",
    "title": "The Quad",
    "description": "Students mention the Quad most often in positive contexts...",
    "image": "./images/quad.jpg",
    "camera": { "latitude": 38.43, "longitude": -78.87, "zoom": 15 },
    "showData": "all_locations"
}
```
